# Pyodide Runtime Demo

This notebook runs entirely in the browser using the Python (Pyodide) kernel bundled with this JupyterLite build.

Use the cells below to inspect the runtime environment and verify WASM execution.

```python
import sys, platform
print(f'Python version: {sys.version.split()[0]}')
print(f'Pyodide platform: {platform.platform()}')
print(f'WASM runtime: {sys.implementation.name}')
```

```python
# Quick computation to show the kernel is responsive
def fibonacci(n):
    if n < 2:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

[fibonacci(n) for n in range(10)]
```
