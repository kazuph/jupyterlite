# Pyodide DOM Magic 🎩✨

This notebook uses the Pyodide kernel to directly manipulate the web page (DOM) of JupyterLite itself.

Click the button to change the UI to a random color!

```python
import js
import random
from IPython.display import display, HTML

def get_random_color():
    return f"#{random.randint(0, 0xFFFFFF):06x}"

def change_color(event):
    # Get the main JupyterLab application element
    app_shell = js.document.getElementById('jp-app-shell')
    if app_shell:
        # Change the background color of a CSS variable
        new_color = get_random_color()
        app_shell.style.setProperty('--jp-layout-color1', new_color)
        print(f"Changed color to {new_color}")

# Create a button
button = js.document.createElement('button')
button.innerHTML = "Change UI Color 🎨"
button.onclick = change_color

# Display the button in the notebook
display(HTML(button.outerHTML))
```
