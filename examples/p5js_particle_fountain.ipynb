# p5.js Example: Particle Fountain ⛲️

This notebook uses the p5.js kernel to create a simple particle fountain. Click anywhere on the canvas to create a new fountain.

```javascript
let emitters = [];

function setup() {
  createCanvas(400, 400);
  // Create an initial emitter in the center
  emitters.push(new Emitter(width / 2, height / 2));
}

function draw() {
  background(10, 10, 20);
  for (let emitter of emitters) {
    emitter.emit(2);
    emitter.update();
    emitter.show();
  }
}

function mousePressed() {
  emitters.push(new Emitter(mouseX, mouseY));
}

class Particle {
  constructor(x, y) {
    this.pos = createVector(x, y);
    this.vel = p5.Vector.random2D().mult(random(1, 4));
    this.acc = createVector(0, 0.05); // Gravity
    this.lifespan = 255;
    this.r = random(255);
    this.g = random(100, 200);
    this.b = random(200, 255);
  }

  finished() {
    return this.lifespan < 0;
  }

  update() {
    this.vel.add(this.acc);
    this.pos.add(this.vel);
    this.lifespan -= 2;
  }

  show() {
    noStroke();
    fill(this.r, this.g, this.b, this.lifespan);
    ellipse(this.pos.x, this.pos.y, 8);
  }
}

class Emitter {
  constructor(x, y) {
    this.position = createVector(x, y);
    this.particles = [];
  }

  emit(num) {
    for (let i = 0; i < num; i++) {
      this.particles.push(new Particle(this.position.x, this.position.y));
    }
  }

  update() {
    for (let particle of this.particles) {
      particle.update();
    }
    this.particles = this.particles.filter(p => !p.finished());
  }

  show() {
    for (let particle of this.particles) {
      particle.show();
    }
  }
}
```
