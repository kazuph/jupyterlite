# p5.js Example: Bouncing Ball Physics 🏀

Click and drag the ball to throw it around. It will bounce off the walls with simple physics.

```javascript
let ball;
let gravity = 0.6;
let friction = -0.9;
let dragging = false;

function setup() {
  createCanvas(400, 400);
  ball = new Ball(width / 2, height / 2, 20);
}

function draw() {
  background(230, 240, 255);
  ball.update();
  ball.checkBoundaryCollision();
  ball.display();
}

function mousePressed() {
  let d = dist(mouseX, mouseY, ball.pos.x, ball.pos.y);
  if (d < ball.radius) {
    dragging = true;
    ball.vel.set(0, 0);
  }
}

function mouseReleased() {
  dragging = false;
}

class Ball {
  constructor(x, y, r) {
    this.pos = createVector(x, y);
    this.vel = createVector(0, 0);
    this.acc = createVector(0, 0);
    this.radius = r;
  }

  update() {
    if (dragging) {
      this.pos.x = mouseX;
      this.pos.y = mouseY;
      // Calculate velocity based on mouse movement
      this.vel.x = mouseX - pmouseX;
      this.vel.y = mouseY - pmouseY;
    } else {
      this.acc.y = gravity;
      this.vel.add(this.acc);
      this.pos.add(this.vel);
    }
  }

  checkBoundaryCollision() {
    if (this.pos.x > width - this.radius) {
      this.pos.x = width - this.radius;
      this.vel.x *= friction;
    } else if (this.pos.x < this.radius) {
      this.pos.x = this.radius;
      this.vel.x *= friction;
    }
    if (this.pos.y > height - this.radius) {
      this.pos.y = height - this.radius;
      this.vel.y *= friction;
    } else if (this.pos.y < this.radius) {
      this.pos.y = this.radius;
      this.vel.y *= friction;
    }
  }

  display() {
    noStroke();
    fill(255, 100, 100);
    ellipse(this.pos.x, this.pos.y, this.radius * 2);
  }
}
```
